# Training the Model using Linear Regression

```python
df.columns
```

```python
X = df[['% Iron Feed', '% Silica Feed', 'Starch Flow', 'Amina Flow',
       'Ore Pulp Flow', 'Ore Pulp pH', 'Ore Pulp Density',
       'Flotation Column 01 Air Flow', 'Flotation Column 02 Air Flow',
       'Flotation Column 03 Air Flow', 'Flotation Column 04 Air Flow',
       'Flotation Column 05 Air Flow', 'Flotation Column 06 Air Flow',
       'Flotation Column 07 Air Flow', 'Flotation Column 01 Level',
       'Flotation Column 02 Level', 'Flotation Column 03 Level',
       'Flotation Column 04 Level', 'Flotation Column 05 Level',
       'Flotation Column 06 Level', 'Flotation Column 07 Level' , '% Iron Concentrate']]
```

```python
y = df['% Silica Concentrate']
```

## Training the Model using Linear Regression

```python
from sklearn.linear_model import LinearRegression
```

```python
model = LinearRegression()
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.33,random_state=101)
```

```python
model.fit(X_train,y_train)
```

# Testing the Model

```python
pred = model.predict(X_test)
```

```python
plt.scatter(y_test,pred,edgecolors='white',color=purple)
```

```python
sns.regplot(x=y_test,y=pred,scatter_kws={"color": red}, line_kws={"color": indigo})
```

```python
sns.displot(y_test - pred)
```

# Performance Matrix

```python
from sklearn import metrics
```

```python
print("Mean Absolute Error = " , metrics.mean_absolute_error(y_test,pred))
print("Mean Squared Error = " , metrics.mean_squared_error(y_test,pred))
print("Root Mean Squared Error = " , np.sqrt(metrics.mean_squared_error(y_test,pred)))
```
